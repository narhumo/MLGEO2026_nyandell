```python
#Now onto matplotlib. Lets make a histogram. First we need some junk data, lets take the following list:
DataF = np.array([14,15,13,21,14,18,19,18,15,18,19,13,9,17])
#Now lets make it into a histogram. The function for a histogram in matplotlib is .hist()
mp.hist(
    [DataF],
    bins=(DataF.max()-DataF.min()),
    histtype='stepfilled',
    label=['Junk Data'],
    edgecolor='black',
)
mp.xlabel('X axis')
mp.ylabel('Y axis')
mp.title('HW1: My Test Histogram')
mp.legend()
mp.tight_layout()
mp.show()

#This is a mediocre histogram with questionable correlation, it's probably fine for an example though.

#To save it is easy, there are many inputs you can do to change what you get but I'm lazy: 
#mp.savefig("HW1 Histogram")
```
